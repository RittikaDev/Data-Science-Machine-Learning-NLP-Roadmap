### Typical EDA Steps — A Practical Checklist

**EDA (Exploratory Data Analysis)** means understanding a dataset, checking its quality, and discovering patterns before modeling. This is a usual order, not a strict rule: revisit earlier steps when you discover something new.

> **Null values and missing values usually mean the same thing in pandas.** `isna()` and `isnull()` perform the same check. Empty strings, spaces, `?`, or sentinel values such as `-999` may also represent missing data, but pandas may not recognize them automatically.

#### 1. Understand the problem and the dataset

- What question are you trying to answer?
- What does one row represent: a person, transaction, day, or something else?
- Understand each column, its units, source, and expected values.
- If modeling, identify the target and inputs available at prediction time.
- Identify ID columns and date/time columns.

#### 2. Load the data and take a first look

- Load the appropriate CSV, Excel, database table, or other source.
- Check the header, delimiter, and parsing settings.
- Look at the first rows, last rows, and a random sample.
- Check the number of rows and columns.

**Useful checks:** `df.head()`, `df.tail()`, `df.sample(5)` (if at least five rows), `df.shape`.

#### 3. Inspect the structure and data types

- Review column names, non-null counts, and types.
- Identify numbers stored as text and dates stored as strings.
- Check whether categorical values are incorrectly treated as measurements, such as numeric IDs.
- Standardize column names if needed; ensure renaming does not create duplicate names.

**Useful checks:** `df.info()`, `df.columns`, `df.dtypes`.

#### 4. Check missing values

- Count missing values in each column and calculate their percentages.
- Display affected rows and investigate patterns: are values missing in a particular group or period?
- Check for disguised missing values such as blank strings or special codes.
- Decide whether to keep, fill, flag, or remove missing values based on their meaning.
- Do not automatically drop all incomplete rows or fill every missing number with zero.

**Useful checks:** `df.isna().sum()`, `df.isna().mean() * 100`, `df[df.isna().any(axis=1)]`.

#### 5. Check duplicate records

- Count exact duplicate rows.
- Also check duplicates using the columns that should uniquely identify a record.
- Investigate before removing: repeated observations can be valid.

**Useful checks:** `df.duplicated().sum()`, `df[df.duplicated(keep=False)]`.

#### 6. Fix inconsistent and invalid values

- Check spelling, capitalization, and spaces in category labels.
- Check impossible values, incorrect units, invalid dates, and inconsistent ranges.
- Check relationships between fields, such as an end date earlier than a start date.
- Convert columns to the correct types after investigating conversion failures.
- Remove irrelevant title/header rows accidentally loaded as observations.
- Recheck shape, missing values, and types after cleaning.

**Useful checks:** `df['column'].unique()`, `df['column'].value_counts(dropna=False)`.

#### 7. Review summary statistics

- For numerical columns, inspect count, mean, median, standard deviation, quartiles, minimum, and maximum.
- For categorical columns, inspect unique counts, frequent values, and rare categories.
- Identify constant columns and columns with unusually many unique values.
- Compare statistics with the real-world meaning and units of each feature.

**Useful checks:** `df.describe()`, `df.describe(include='all')`, `df.nunique()`.

#### 8. Explore each variable separately — univariate analysis

- Numerical variables: use histograms and box plots to inspect distribution, spread, skewness, gaps, and extreme values.
- Categorical variables: use frequency tables and count/bar plots.
- Do not assume every numerical column should follow a normal distribution.
- Treat numeric category codes as categories when appropriate.

#### 9. Investigate outliers

- Flag unusually high or low values using plots, domain limits, or methods such as the IQR rule.
- Decide whether each extreme value is an error or a valid observation.
- Correct, retain, transform, cap, or remove values only with a documented reason.
- Outliers are not automatically mistakes; automatic removal can discard useful information.

#### 10. Examine the target, if there is one

- Regression target: inspect its range, distribution, skewness, missing values, and outliers.
- Classification target: inspect class counts and percentages for imbalance.
- Check that labels are valid and consistent.
- Missing targets usually cannot be used directly for supervised training.

#### 11. Explore relationships between variables — bivariate analysis

- Numerical vs numerical: scatter plots and suitable correlations.
- Categorical vs numerical: grouped summaries, box plots, or violin plots.
- Categorical vs categorical: cross-tabulations and grouped/stacked bar charts.
- If modeling, explore how each input relates to the target.
- Correlation does not prove causation, and low linear correlation does not rule out a curved relationship.

**Useful checks:** `df.groupby('category')['value'].agg(['count', 'mean', 'median'])`, `pd.crosstab(df['category_a'], df['category_b'])`.

#### 12. Explore combined relationships — multivariate analysis

- Use selected pair plots, correlation heatmaps, or plots colored by a third variable.
- Check whether relationships differ by region, category, or another relevant group.
- Investigate highly correlated predictors and potential redundancy.
- Do not automatically remove a feature just because it exceeds a correlation threshold.
- Correlations of arbitrary numeric category codes may be misleading.

#### 13. Check time patterns and dataset coverage, when relevant

- Sort dates and look for trends, seasonality, missing periods, and repeated timestamps.
- Check whether some regions, groups, or periods are underrepresented.
- Consider whether nearby observations or repeated records from the same person are dependent.
- Ask whether this dataset represents the population or future situations you care about.

#### 14. Record preprocessing and feature-engineering decisions

- Identify useful date components, transformations, category groupings, or new features.
- Decide which identifiers, redundant fields, or unavailable-at-prediction-time columns to exclude.
- Check for target leakage: an input must not reveal information that would only be known after the outcome.
- Encoding and scaling may be needed for modeling, but they are not mandatory for every EDA.

#### 15. Summarize findings and verify the cleaned data

- Record important patterns, data-quality issues, limitations, and unanswered questions.
- Document every cleaning decision and how many records it affected.
- Recheck row counts, types, missing values, duplicates, and valid ranges.
- Preserve the raw data and save a separate cleaned dataset when appropriate.
- Decide the next analysis or modeling step from the evidence.

#### Important boundary when preparing for machine learning

Perform basic schema and integrity checks first, then reserve the test set **before detailed model-guiding exploration or data-driven preprocessing**. Use training data for feature selection, learned outlier thresholds, imputation statistics, scaling, and other fitted transformations. Apply those same learned rules to validation/test data. During cross-validation, fit preprocessing separately within each training fold. Use time-aware or group-aware splits when the observations require them.

#### Quick order to remember

**Understand → Load → Inspect structure → Check missing values → Check duplicates → Fix inconsistencies → Summarize → Explore distributions → Investigate outliers → Inspect target → Explore relationships → Check time/groups → Plan features → Document and verify.**
