```python
import numpy as np

data = np.array([10, 12, 13, 14, 15, 16, 18, 20, 100])
```

```python
np.mean(data)
np.median(data)
np.std(data)
np.var(data)
```

```python
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Standard Deviation:", np.std(data))
print("Variance:", np.var(data))
```

```python
data_without_outlier = np.array([10, 12, 13, 14, 15, 16, 18, 20])
```

```python
np.mean(data_without_outlier)
np.median(data_without_outlier)
np.std(data_without_outlier)
np.var(data_without_outlier)
```

```python
print("Mean:", np.mean(data_without_outlier))
print("Median:", np.median(data_without_outlier))
print("Standard Deviation:", np.std(data_without_outlier))
print("Variance:", np.var(data_without_outlier))
```

```python
# OBSERVATION:
# THE MEAN IS SIGNIFICANTLY AFFECTED BY THE OUTLIER (100), WHILE THE MEDIAN REMAINS RELATIVELY STABLE. THE STANDARD DEVIATION AND VARIANCE ARE ALSO INFLUENCED BY THE OUTLIER, INDICATING THAT THEY ARE SENSITIVE TO EXTREME VALUES IN THE DATASET.
# STANDARD DEVIATION MEASURES HOW FAR VALUES ARE SPREAD FROM THE MEAN.
```
